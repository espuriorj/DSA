```python
# Import necessary libraries
import numpy as np
import pandas as pd
import seaborn as sns

# Load the dataset
data = pd.read_csv('/content/drive/MyDrive/lr.csv')

# Rename columns for clarity
data.columns = ['X','Y']

# Display the dataset
print(data)

# Access an element in the dataset (example: row 2, column 1)
print(data.iloc[1,0])

# Define parameters for linear regression
epochs = 100
lr = 0.1
m = 0
c = 0
batch_size = 20

# Define a function to calculate Mean Squared Error (MSE) for a random batch
import random

def mse(batch_size=99, data=data, m=m, c=c):
    # Generate a random sample of indices for the batch
    x = random.sample([i for i in range(0, 99)], batch_size)
    
    # Calculate the errors for the selected batch
    errors = [(m * data.iloc[i, 0] + c) - data.iloc[i, 1] for i in x]
    
    return errors, batch_size, x, m, c

# Apply Min-Max scaling to the 'X' column
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

x1 = np.array(data['X']).reshape(-1, 1)
scaler.fit(x1)
l1 = scaler.transform(x1)

x1 = l1.reshape(1, -1)[0]
data['X'] = x1

# Display the scaled dataset
print(data)

# Access an element in the scaled dataset (example: row 99, column 1)
print(data.iloc[98, 1])

# Initialize variables for linear regression
m = 0
c = 0
lr = 0.1

# Display the values of m and c
print(m)
print(c)

# Display the dataset
print(data)

# Display an element in the dataset (example: row 4, column 2)
print(data.iloc[3, 1])

# Initialize a list to store the loss for each epoch
loss = []

# Perform gradient descent for linear regression
for i in range(0, epochs):
    # Select a random sample of one data point for each epoch
    x = random.sample([i for i in range(0, 99)], 1)
    
    # Initialize loss_m and loss_c
    loss_m = 0
    loss_c = 0
    
    # Calculate the loss for the selected data point
    for j in x:
        loss_m += (m * data.iloc[j, 0] + c - data.iloc[j, 1]) * data.iloc[j, 0]
        loss_c += m * data.iloc[j, 0] + c - data.iloc[j, 1]
    
    # Update m and c using gradient descent
    m = m - lr * loss_m
    c = c - lr * loss_c
    
    # Append the squared loss for the selected data point to the loss list
    loss.append((m * data.iloc[j, 0] + c - data.iloc[j, 1]) ** 2)

# Display the final values of m and c
print(m)
print(c)

# Extract the last 100 losses for plotting
loss = loss[len(loss) - 100:len(loss)]

# Create a line plot to visualize the loss over the last 100 epochs
line = [y for y in range(0, 100)]
sns.lineplot(x=line, y=loss)

# Display the length of the loss list
print(len(loss))

# Display the values of the last 100 losses
print(loss)

# Plot the original dataset using a scatter plot
sns.scatterplot(data=data, x='X', y='Y')

# Generate a line plot to visualize the linear regression line
x_line = np.linspace(min(data['X']), max(data['X']), 99)
y_line = m * x_line + c
plt.plot(x_line, y_line, color='red')

# Generate another line plot for the linear regression line (repeated code)
x_line = np.linspace(min(data['X']), max(data['X']), 99)
y_line = m * x_line + c
plt.plot(x_line, y_line, color='red')
```
